```python
df=pd.read_csv("Food_Delivery_Time_Prediction.csv")
df
```

```python
df.drop(["Order_ID","Order_Date"],axis=1,inplace=True)
```

```python
x=df.drop('Time_taken_min',axis=1)
x
```

```python
y=df['Time_taken_min']
y
```

```python
cat_fet= x.select_dtypes(include=['object']).columns.tolist()
cat_fet
```

```python
num_fet= x.select_dtypes(exclude=['object']).columns.tolist()
num_fet
```

```python
nom_cols=x[['Day_of_Week','Weather','Pickup_Zone','Dropoff_Zone','Cuisine_Type']]
nom_cols
```

```python
ord_features=['Restaurant_Load','Delivery_Distance_Category','Traffic_Level','Delivery_Priority']
ord_features
```

```python
nom_features=nom_cols.columns
nom_features
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler,OneHotEncoder,OrdinalEncoder
```

```python
num_pipe=Pipeline(steps=[("imp",SimpleImputer(strategy='median')),
                 ("scaler",StandardScaler())])
num_pipe
```

```python
nom_pipe=Pipeline(steps=[('imp',SimpleImputer(strategy='most_frequent')),
                   ('onehot',OneHotEncoder(sparse_output=False,drop='first'))])
nom_pipe
```

```python
ord_pipe=Pipeline(steps=[('imp',SimpleImputer(strategy='most_frequent')),
                         ('ord',OrdinalEncoder(categories=[
                             ['Low','Medium','High'],
                             ['Short','Medium','Long'],
                             ['Low','Moderate','High','Severe'],
                             ['Normal','Priority','VIP']]))])
ord_pipe
```

```python
from sklearn.compose import ColumnTransformer
```

```python
full_pipe=ColumnTransformer(transformers=[
    ('num',num_pipe,num_fet),
    ('nom',nom_pipe,nom_features),
    ('ord',ord_pipe,ord_features)])
full_pipe
```

```python
from sklearn.linear_model import LinearRegression
linear=Pipeline(steps=[('pipe',full_pipe),
                       ('model',LinearRegression())])
linear
```

```python
import os
print(os.getcwd())
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
linear.fit(x_train,y_train)
```

```python
print(linear.score(x_train,y_train))
linear.score(x_test,y_test)
```
